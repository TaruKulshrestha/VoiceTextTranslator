```python
import speech_recognition as sr
import pyttsx3
from gtts import gTTS
import os
from googletrans import Translator

# Initialize the recognizer
r = sr.Recognizer()

# Initialize the engine for text-to-speech
engine = pyttsx3.init()

# Dictionary of supported languages and their language codes
supported_languages = {
    "Hindi": "hi",
    "Tamil": "ta",
    "Bengali": "bn",
    "Marathi": "mr"
}

# Function to convert text to speech (for confirmation and errors)
def SpeakText(command):
    engine.say(command)
    engine.runAndWait()

# Function to select language
def select_language():
    print("Please select a language:")
    for idx, language in enumerate(supported_languages.keys(), 1):
        print(f"{idx}. {language}")
    choice = int(input("Enter the number of your choice: "))
    language_name = list(supported_languages.keys())[choice - 1]
    language_code = supported_languages[language_name]
    return language_code

# Function to read text from a file
def read_text_from_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        return file.read()

# Function to translate text to the selected language
def translate_text(text, target_language):
    translator = Translator()
    translated = translator.translate(text, dest=target_language)
    return translated.text

# Main function to handle user choice
def main():
    print("Select an option:")
    print("1. Speech to Text")
    print("2. Text to Speech")
    choice = int(input("Enter the number of your choice: "))

    if choice == 1:
        # Set the language code for speech recognition
        language_code = select_language()

        # Loop infinitely for user to speak
        while True:
            try:
                # Use the microphone as source for input
                with sr.Microphone() as source2:
                    # Adjust the energy threshold based on the surrounding noise level
                    r.adjust_for_ambient_noise(source2, duration=0.2)
                    
                    print("Listening...")
                    # Listens for the user's input
                    audio2 = r.listen(source2)
                    
                    print("Recognizing...")
                    # Using Google to recognize audio
                    MyText = r.recognize_google(audio2, language=language_code)
                    MyText = MyText.lower()

                    print("Did you say: ", MyText)

                    # Check for exit command
                    if "exit" in MyText:
                        print("Exiting program...")
                        break
                    
            except sr.RequestError as e:
                print("Could not request results; {0}".format(e))
                
            except sr.UnknownValueError:
                print("Unknown error occurred. Please speak clearly.")
    
    elif choice == 2:
        # Set the language code for TTS
        language_code = select_language()

        # File path to the text file
        file_path = r"C:\Users\Admin\Desktop\mini project\text_to_speech.txt"

        # Read text from file
        text_to_speak = read_text_from_file(file_path.strip())

        # Translate text to the selected language
        translated_text = translate_text(text_to_speak, language_code)

        # Convert text to speech using gTTS
        tts = gTTS(text=translated_text, lang=language_code, slow=False)
        tts.save("output.mp3")

        # Play the converted speech
        os.system("start output.mp3" if os.name == "nt" else "afplay output.mp3")

        # Optionally, remove the file after playing
        os.remove("output.mp3")
    else:
        print("Invalid choice. Please restart the program and choose a valid option.")

if __name__ == "__main__":
    main()

```
